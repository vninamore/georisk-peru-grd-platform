# GeoRisk Peru — análisis reproducible de un escenario GRD
Este notebook usa datos **sintéticos** para revisar la distribución de peligro, exposición, vulnerabilidad y riesgo.

```python
from pathlib import Path
import pandas as pd

DATA = Path('../data/sample/vulnerability_indicators.csv')
df = pd.read_csv(DATA)
df.head()
```

```python
weights = {'physical': .35, 'social': .30, 'economic': .20, 'environmental': .15}
df['vulnerability'] = sum(df[column] * weight for column, weight in weights.items())
df['risk'] = df['hazard'] * df['exposure'] * df['vulnerability']
df.sort_values('risk', ascending=False)[['unit_id', 'hazard_type', 'risk']]
```

## Interpretación
El ranking orienta la priorización analítica. Antes de usarlo para decisiones se deben calibrar pesos y umbrales, validar fuentes y contrastar resultados con trabajo de campo y juicio experto.
